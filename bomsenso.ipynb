```python
# ============================================================
# @title CÉLULA 6: Motor LazyGraphRAG
# Objetivo: Ingerir os componentes + vulnerabilidades do VDR
# ============================================================

!pip install networkx==3.2.1 --quiet

import os
import json
import networkx as nx
from langchain_core.tools import tool

print("="*70)
print(" 🕸️ INICIANDO MOTOR LAZY GRAPHRAG (INGESTÃO COMPLETA)")
print("="*70)

VDR_FILE = "artefatos/depscan-universal.vdr.json"
grafo_conhecimento = nx.DiGraph()

def construir_grafo_completo():
    if not os.path.exists(VDR_FILE):
        print(f"[!] Arquivo '{VDR_FILE}' não encontrado.")
        return

    try:
        with open(VDR_FILE, "r", encoding="utf-8") as f:
            vdr_data = json.load(f)

        componentes = vdr_data.get("components", [])
        vulnerabilidades = vdr_data.get("vulnerabilities", [])

        # 1. Ingestão completa dos Componentes
        comps_adicionados = 0
        for comp in componentes:
            bom_ref = comp.get("bom-ref") or comp.get("purl") or comp.get("name")
            nome = comp.get("name", "Desconhecido")
            versao = comp.get("version", "N/A")
            tipo = comp.get("type", "unknown")

            if bom_ref:
                grafo_conhecimento.add_node(
                    bom_ref,
                    label="Componente",
                    name=nome,
                    version=versao,
                    type=tipo
                )
                comps_adicionados += 1

        # 2. Ingestão completa das Vulnerabilidades e conexões
        vulns_adicionadas = 0
        for vuln in vulnerabilidades:
            cve_id = vuln.get("id")
            if not cve_id:
                continue

            severidade = "UNKNOWN"
            ratings = vuln.get("ratings", [])
            if ratings:
                severidade = ratings[0].get("severity", "UNKNOWN").upper()

            grafo_conhecimento.add_node(
                cve_id,
                label="Vulnerabilidade",
                severity=severidade
            )
            vulns_adicionadas += 1

            # Conecta aos componentes afetados
            afetados = vuln.get("affects", [])
            for af in afetados:
                comp_ref = af.get("ref")
                if comp_ref and grafo_conhecimento.has_node(comp_ref):
                    grafo_conhecimento.add_edge(comp_ref, cve_id, relation="HAS_VULNERABILITY")

        print(f"✔ Grafo de Conhecimento Totalmente Sincronizado!")
        print(f"  Componentes Indexados: {comps_adicionados}")
        print(f"  Vulnerabilidades Indexadas: {vulns_adicionadas}")
        print(f"  Total de Nós no Grafo: {grafo_conhecimento.number_of_nodes()}")
        print(f"  Total de Arestas de Impacto: {grafo_conhecimento.number_of_edges()}")

    except Exception as e:
        print(f"[!] Erro crítico: {e}")

construir_grafo_completo()

# =====================================================================
# FERRAMENTA DE CONSULTA SOB DEMANDA (Lazy RAG - Busca Semântica)
# =====================================================================
@tool
def lazy_graph_retriever(query: str = "ALL") -> str:
    """
    Ferramenta de busca em Grafo para o Comitê de IA.
    Busca vulnerabilidades baseando-se nas palavras-chave contidas na query.
    """
    query_upper = query.upper()
    
    # 1. Identifica quais severidades o Agente mencionou na frase
    severidades_buscadas = [sev for sev in ["CRITICAL", "HIGH", "MEDIUM", "LOW"] if sev in query_upper]
    
    # 2. Fallback: Se ele mandar "ALL" ou uma frase genérica sem severidade clara, trazemos as mais graves
    if "ALL" in query_upper or not severidades_buscadas:
        severidades_buscadas = ["CRITICAL", "HIGH"]
        
    resultado = []

    for no_id, dados in grafo_conhecimento.nodes(data=True):
        if dados.get("label") == "Vulnerabilidade":
            # 3. Verifica se a severidade deste nó está na lista das buscadas
            if dados.get("severity") in severidades_buscadas:
                pacotes_afetados = list(grafo_conhecimento.predecessors(no_id))
                nomes_pacotes = []
                for p_ref in pacotes_afetados:
                    d_pacote = grafo_conhecimento.nodes[p_ref]
                    nomes_pacotes.append(f"{d_pacote.get('name')} (v. {d_pacote.get('version')})")

                if nomes_pacotes:
                    resultado.append(f"- {no_id} [{dados.get('severity')}]: afeta {', '.join(nomes_pacotes)}")
                else:
                    resultado.append(f"- {no_id} [{dados.get('severity')}]")

    if not resultado:
        return f"Auditoria Limpa: Nenhuma vulnerabilidade encontrada para os filtros: {severidades_buscadas}."

    # 4. Formatação de saída (limite de 50 para não estourar contexto do Analista)
    resumo_str = f"Resumo de Risco VDR ({', '.join(severidades_buscadas)}): {len(resultado)} ocorrências mapeadas.\n"
    resumo_str += "\n".join(resultado[:50]) 

    if len(resultado) > 50:
        resumo_str += f"\n... (e mais {len(resultado) - 50} vulnerabilidades omitidas)."

    return resumo_str

print("✔ Ferramenta '@tool lazy_graph_retriever' completa e otimizada com parser de substrings!")
print("="*89)
```

```python
# =================================================================================
# @title CÉLULA 7: Orquestração Zero Trust (Artifact-Mediated Handoff & HiTL)
# Objetivo: Comitê dinâmico com passagem de referência de arquivos e Atestação PQC
# =================================================================================
import io
import os
import time
import zipfile
import json
import hashlib
import oqs
import re
from datetime import datetime, timezone
from typing import TypedDict
from IPython.display import display, HTML, clear_output
import ipywidgets as widgets
from langgraph.graph import StateGraph, END

# =====================================================================
# 1. Configuração de Diretórios e Persistência (Sobrescrita)
# =====================================================================
DIR_RELATORIOS = "relatorios_v100"
os.makedirs(DIR_RELATORIOS, exist_ok=True)

def salvar_artefato(nome_base, conteudo):
    """Salva o artefato com nomenclatura sem numerais no início (sobrescrevendo o anterior)"""
    filepath = os.path.join(DIR_RELATORIOS, f"{nome_base}_1.md")
    with open(filepath, "w", encoding="utf-8") as f:
        f.write(conteudo)
    return filepath

def ler_artefato(filepath):
    """Lê o artefato físico do disco via referência integral"""
    with open(filepath, "r", encoding="utf-8") as f:
        return f.read()

def extrair_veredito(filepath, fallback_chars=1200):
    """Lê o artefato físico e extrai a conclusão final. Se o LLM falhar, o Python constrói a frase com Regex."""
    with open(filepath, "r", encoding="utf-8") as f:
        texto = f.read()
        texto_upper = texto.upper()
        
        # 1. Tenta achar a tag oficial
        if "=== VEREDITO ===" in texto_upper:
            idx = texto_upper.find("=== VEREDITO ===") + len("=== VEREDITO ===")
            veredito = texto[idx:].strip()
            return veredito.replace("--- FIM DO FORMATO OBRIGATÓRIO ---", "").strip()
            
        # 2. Tenta achar o início da frase caso ele tenha esquecido a tag
        elif "APÓS ANALISAR" in texto_upper:
            idx_orig = texto_upper.find("APÓS ANALISAR")
            veredito = texto[idx_orig:].strip()
            return veredito.replace("--- FIM DO FORMATO OBRIGATÓRIO ---", "").strip()
            
        # 3. Escudo Anti-Preguiça Corrigido: Extrai apenas o total de ameaças severas
        try:
            with open(os.path.join(DIR_RELATORIOS, "contexto_rag_1.md"), "r", encoding="utf-8") as frag:
                texto_rag = frag.read()
            
            m1 = re.search(r'(\d+)\s+ocorrências mapeadas', texto_rag, re.IGNORECASE)
            mapeadas = int(m1.group(1)) if m1 else 0
            
            if mapeadas > 0:
                return f"Após analisar o Grafo de Conhecimento, identifiquei {mapeadas} ameaças severas (CRITICAL/HIGH), das quais priorizei as 10 mais críticas. Sugiro que o comitê superior avalie o relatório salvo em disco para validação e decisão final."
        except Exception:
            pass
            
        return "Após analisar as vulnerabilidades indexadas no Grafo de Conhecimento, identifiquei as ameaças severas e priorizei o Top 10 crítico. Sugiro que o comitê superior avalie o relatório salvo em disco para validação e decisão final."

# =====================================================================
# 2. Definição de Estado do Grafo (Control Plane de Referências)
# =====================================================================
class ComiteState(TypedDict):
    relatorio_bruto_ref: str
    parecer_analista_ref: str
    revisao_auditor_ref: str
    decisao_final_ref: str
    auditor_ignorado: bool

DATA_ATUAL = datetime.now().strftime("%d de %B de %Y")
PERFIL_HW = LLM_CONFIG.get("perfil", "Fusca").lower() if 'LLM_CONFIG' in globals() else "fusca"

# =====================================================================
# 3. Instanciação dos Motores (Dimensionamento de Contexto via VRAM)
# =====================================================================
def instanciar_agente(config_persona, temperatura):
    engine = config_persona["engine"]
    modelo = config_persona["modelo"]
    
    # Política de contexto derivada da capacidade de VRAM do perfil
    vram_gb = LLM_CONFIG.get("vram_gb", 4.0)
    
    if vram_gb <= 6.5: num_ctx = 4096
    elif vram_gb <= 8.5: num_ctx = 8192
    elif vram_gb <= 12.5: num_ctx = 12288
    elif vram_gb <= 16.5: num_ctx = 16384
    elif vram_gb <= 20.5: num_ctx = 20480
    elif vram_gb <= 24.5: num_ctx = 24576
    elif vram_gb <= 32.5: num_ctx = 32768
    elif vram_gb <= 40.5: num_ctx = 40960
    elif vram_gb <= 48.5: num_ctx = 49152
    elif vram_gb <= 64.5: num_ctx = 65536
    elif vram_gb <= 80.5: num_ctx = 81920
    elif vram_gb <= 96.5: num_ctx = 98304
    elif vram_gb <= 141.5: num_ctx = 131072
    elif vram_gb <= 180.5: num_ctx = 163840
    elif vram_gb <= 192.5: num_ctx = 196608
    elif vram_gb <= 288.5: num_ctx = 262144
    else: num_ctx = 262144 # Absolute Resolve (256K+)

    if engine == "huggingface":
        from langchain_huggingface import HuggingFacePipeline
        return HuggingFacePipeline.from_model_id(
            model_id=modelo, 
            task="text-generation", 
            pipeline_kwargs={
                "temperature": temperatura,
                "max_new_tokens": min(4096, num_ctx // 4) 
            }
        )
        
    elif engine == "ollama":
        from langchain_ollama import ChatOllama
        return ChatOllama(
            model=modelo, 
            temperature=temperatura, 
            base_url="http://ollama:11434", 
            num_ctx=num_ctx
        )

CONF_ANALISTA = LLM_CONFIG["hierarquia"]["Analista"]
CONF_SUPERVISOR = LLM_CONFIG["hierarquia"]["Supervisor"]
CONF_AUDITOR = LLM_CONFIG["assurance"]["Auditor Interno"]

llm_analista = instanciar_agente(CONF_ANALISTA, 0.1)
llm_auditor = instanciar_agente(CONF_AUDITOR, 0.4)
llm_supervisor = instanciar_agente(CONF_SUPERVISOR, 0.1)

# =====================================================================
# 4. Nós do LangGraph (Handoff de Artefatos Físicos & Dinâmicos)
# =====================================================================
def node_analyst(state: ComiteState):
    print("\n🔍 [RAG] Conectando ao Grafo de Conhecimento para extração de vulnerabilidades...")
    
    if 'lazy_graph_retriever' not in globals():
        erro_msg = "[FALHA CRÍTICA] Motor RAG não encontrado. Verifique as células anteriores."
        print(f"❌ {erro_msg}")
        raise RuntimeError(erro_msg)
        
    query_rag = "vulnerability CVE CRITICAL HIGH"
    contexto_rag = lazy_graph_retriever.invoke({"query": query_rag})
    texto_rag_str = str(contexto_rag)
    caminho_rag = salvar_artefato("contexto_rag", texto_rag_str)
    
    papel_analista = CONF_ANALISTA.get("descricao", "Analista de Segurança Nível 1")
    
    # --- CÁLCULO VIA PYTHON REGEX CORRIGIDO ---
    match_mapeadas = re.search(r'(\d+)\s+ocorrências mapeadas', texto_rag_str, re.IGNORECASE)
    mapeadas = int(match_mapeadas.group(1)) if match_mapeadas else 0
    # --------------------------------
    
    # ---------------------------------------------------------
    # PROMPT DO ANALISTA: Frase Mastigada e Honesta
    # ---------------------------------------------------------
    prompt = f"""[DATA DO SISTEMA: {DATA_ATUAL}]
    Atue como: {papel_analista}. Responda em Português (PT-BR).
    PREENCHA O TEMPLATE ABAIXO OBRIGATORIAMENTE ATÉ O FINAL:

    --- INÍCIO DO FORMATO OBRIGATÓRIO ---
    ### 🚨 Top 10 Vulnerabilidades
    (Liste apenas as 10 piores falhas extraídas, citando o CVE e componente afetado em tópicos)

    ### 🛠️ Recomendações Operacionais
    (Escreva 2 tópicos curtos sugerindo mitigação)

    === VEREDITO ===
    Após analisar o Grafo de Conhecimento, identifiquei {mapeadas} ameaças severas (CRITICAL/HIGH), das quais priorizei as 10 mais críticas. Sugiro que o comitê superior avalie o relatório salvo em disco para validação e decisão final.
    --- FIM DO FORMATO OBRIGATÓRIO ---
    
    Dados Brutos (RAG):
    {texto_rag_str}"""
    
    print(f"⏰ [LLM] Motor Analista DESPERTADO (Perfil: {PERFIL_HW.upper()}). Processando dados via template Júnior...")
    
    resposta = llm_analista.invoke(prompt)
    conteudo = resposta.content if hasattr(resposta, 'content') else str(resposta)
    
    caminho_analista = salvar_artefato("parecer_analista", conteudo)
    print(f"✅ [IO] Parecer do Analista materializado em: {caminho_analista}")
    print("💤 [LLM] Motor Analista SUSPENSO (Handoff concluído).")
    
    return {"relatorio_bruto_ref": caminho_rag, "parecer_analista_ref": caminho_analista}

def node_auditor(state: ComiteState):
    print("\n🕵️ [LLM] Lendo referências e iniciando revisão crítica do Auditor. Aguarde...")
    texto_analista = ler_artefato(state["parecer_analista_ref"])
    papel_auditor = CONF_AUDITOR.get("descricao", "Auditor Interno de Segurança")
    
    # ---------------------------------------------------------
    # PROMPT DO AUDITOR: Enxuto, Direto e Sem Metalinguagem
    # ---------------------------------------------------------
    prompt = f"""[DATA DO SISTEMA: {DATA_ATUAL}]
    Atue como: {papel_auditor}. Responda estritamente em Português (PT-BR).
    O Analista submeteu o parecer abaixo. Sua tarefa é analisar se há lacunas metodológicas ou falta de versões específicas.

    Parecer submetido:
    {texto_analista}

    INSTRUÇÃO DE SAÍDA (OBRIGATÓRIA):
    1. Aponte em até 2 tópicos curtos as falhas ou lacunas encontradas.
    2. Encerre obrigatoriamente com a linha exata: === VEREDITO DA AUDITORIA === seguida por sua conclusão técnica sobre o rigor do documento."""
    
    resposta = llm_auditor.invoke(prompt)
    conteudo = resposta.content if hasattr(resposta, 'content') else str(resposta)
    
    # Validação dinâmica em Python: Se o LLM cortar a resposta e omitir o veredito,
    # damos uma segunda chance pedindo um complemento focado apenas na conclusão,
    # mantendo a esteira 100% dinâmica.
    if "=== VEREDITO DA AUDITORIA ===" not in conteudo.upper():
        print("⚠️ [Auditor] Resposta truncada detectada. Solicitando conclusão dinâmica...")
        complemento_prompt = f"Com base na análise anterior, adicione apenas a linha final obrigatória:\n=== VEREDITO DA AUDITORIA ===\n[Sua conclusão técnica aqui]"
        resp_comp = llm_auditor.invoke(complemento_prompt)
        conteudo += "\n\n" + (resp_comp.content if hasattr(resp_comp, 'content') else str(resp_comp))

    caminho_auditor = salvar_artefato("revisao_auditor", conteudo)
    print(f"✅ [IO] Revisão do Auditor materializada em: {caminho_auditor}")
    return {"revisao_auditor_ref": caminho_auditor}

def node_supervisor(state: ComiteState):
    print("\n⏰ [LLM] Motor Supervisor DESPERTADO (CISO). Iniciando deliberação executiva...")
    ignorado = state.get("auditor_ignorado", False)
    
    # O CISO lê o resumo do Analista (a extração está correta aqui)
    texto_analista = extrair_veredito(state["parecer_analista_ref"], fallback_chars=2500)
    papel_supervisor = CONF_SUPERVISOR.get("descricao", "Supervisor do Comitê de Governança (CISO)")
    
    # ---------------------------------------------------------
    # PROMPT DO SUPERVISOR: Lógica Estrita Zero Trust (Sem Vendas)
    # ---------------------------------------------------------
    if ignorado:
        contexto_auditoria = """[INCONFORMIDADE DE GOVERNANÇA]: O operador humano omitiu a etapa obrigatória do Auditor Interno.
        REGRA ZERO TRUST: Você é OBRIGADO a BLOQUEAR este artefato. 
        Explique o risco operacional da ausência de revisão dual e decrete o bloqueio de forma técnica."""
    else:
        # CORREÇÃO CRÍTICA: Lemos o documento INTEGRAL do Auditor, com todas as críticas.
        texto_auditor = ler_artefato(state["revisao_auditor_ref"])
        contexto_auditoria = f"""Revisão Integral do Auditor Interno: 
        {texto_auditor}
        
        REGRA ZERO TRUST DE DELIBERAÇÃO: 
        1. Leia a Revisão do Auditor acima com extrema atenção. Se o texto contiver termos como "Lacuna", "Falhas metodológicas", criticar a ausência de versões exatas ou apontar falta de evidências, você é OBRIGADO a decretar o status como BLOQUEADO.
        2. ATENÇÃO: Nunca assuma que os apontamentos ou exigências do Auditor são evidências de que o ambiente está seguro. Eles são a prova de que o documento do Analista é fraco.
        3. Para aprovar, o Auditor deve ter atestado explicitamente a solidez do documento."""

    prompt = f"""[DATA DO SISTEMA: {DATA_ATUAL}]
    Atue estritamente como: {papel_supervisor}.
    REQUISITO CRITICO: Retorne APENAS o texto do parecer executivo. Sem metalinguagem.
    
    Síntese Analítica Original (Analista):
    {texto_analista}
    
    {contexto_auditoria}
    
    Instrução Final: Inicie o seu texto declarando OBRIGATORIAMENTE "[DECISÃO: APROVADO]" ou "[DECISÃO: BLOQUEADO]", seguido de um parágrafo rigoroso justificando a sua decisão citando os apontamentos exatos feitos pelo Auditor."""
    
    resposta = llm_supervisor.invoke(prompt)
    conteudo = resposta.content if hasattr(resposta, 'content') else str(resposta)
    
    caminho_supervisor = salvar_artefato("veredito_supervisor", conteudo)
    print(f"✅ [IO] Veredito do Supervisor materializado em: {caminho_supervisor}")
    print("💤 [LLM] Motor Supervisor SUSPENSO. Esteira finalizada.")
    
    return {"decisao_final_ref": caminho_supervisor}
    
# =====================================================================
# 5. Módulo HiTL Dinâmico (Human-in-the-Loop)
# =====================================================================
estado_global = ComiteState(relatorio_bruto_ref="", parecer_analista_ref="", revisao_auditor_ref="", decisao_final_ref="", auditor_ignorado=False)
saida_interativa = widgets.Output()
display(saida_interativa)

with saida_interativa:
    print("="*80)
    print(f" 🚀 INICIANDO ESTEIRA (ARTIFACT HANDOFF) | Perfil: {PERFIL_HW.upper()}")
    print("="*80)
    estado_global.update(node_analyst(estado_global))
    print("\n🧑‍💼 RESUMO EXECUTIVO DO ANALISTA:")
    print("-" * 80)
    print(extrair_veredito(estado_global["parecer_analista_ref"]))
    print("-" * 80)
    print(f"📂 Relatório completo gravado em disco: {estado_global['parecer_analista_ref']}")

btn_sim_auditor = widgets.Button(description='🕵️ Acionar Auditor', button_style='success', layout=widgets.Layout(width='200px'))
btn_nao_auditor = widgets.Button(description='⚠ Pular Auditor (Risco)', button_style='danger', layout=widgets.Layout(width='200px'))
ui_decisao = widgets.HBox([btn_sim_auditor, btn_nao_auditor])

def finalizar_esteira():
    if not estado_global["auditor_ignorado"]:
        html_pqc = """
        <div style="margin-top: 20px; padding: 15px; border-left: 4px solid #9c27b0; background-color: #1a1a1a;">
            <h4 style="margin:0; color: #e1bee7;">🔐 Atestação Pós-Quântica (FIPS 204)</h4>
            <p style="font-size: 13px; color: #ccc;">Gerar envelope criptográfico com arquivos .sig físicos:</p>
        </div>
        """
        dropdown_pqc = widgets.Dropdown(options=['ML-DSA-44', 'ML-DSA-65', 'ML-DSA-87'], value='ML-DSA-65', description='FIPS 204:')
        btn_zip = widgets.Button(description='📦 Gerar Envelope PQC', button_style='primary', icon='lock')
        
        def gerar_zip(b):
            with saida_interativa:
                output_dir = os.path.join("assets", "pqc_signatures")
                os.makedirs(output_dir, exist_ok=True)
                utc_timestamp = datetime.now(timezone.utc).strftime("%Y%m%dT%H%M%SZ")
                
                # Nome do envelope final que será lido pelo validador
                zip_filename = f"aimlbom_fips204_{utc_timestamp}.zip"
                zip_filepath = os.path.join(output_dir, zip_filename)
                algoritmo_selecionado = dropdown_pqc.value
                
                # 1. Cria o Payload Canônico (artefato.zip) em memória
                payload_buffer = io.BytesIO()
                with zipfile.ZipFile(payload_buffer, 'w', zipfile.ZIP_DEFLATED) as temp_zip:
                    
                    # CORREÇÃO: Apontando para o diretório correto
                    pasta_artefatos = "artefatos"
                    
                    for f_json in ["aibom.json", "cbom.json", "obom.json", "saasbom.json", "sbom.json"]:
                        caminho_completo = os.path.join(pasta_artefatos, f_json)
                        
                        # Verifica se o arquivo existe dentro da pasta 'artefatos'
                        if os.path.exists(caminho_completo): 
                            # O 'arcname=f_json' garante que fique na raiz do zip interno
                            temp_zip.write(caminho_completo, arcname=f_json)
                            print(f"[DEBUG] Anexado ao Payload: {f_json}") # Log visual extra
                            
                    for dir_name in ["reports", "relatorios_v100"]:
                        if os.path.exists(dir_name):
                            for root, _, files in os.walk(dir_name):
                                for file in files:
                                    path = os.path.join(root, file)
                                    temp_zip.write(path, arcname=os.path.relpath(path, start="."))
                
                # Extrai os bytes exatos que serão assinados
                payload_bytes = payload_buffer.getvalue()
                
                # Calcula hashes para os metadados (garantindo rastreabilidade adicional)
                sha256_hash = hashlib.sha256(payload_bytes).hexdigest()
                sha3_256_hash = hashlib.sha3_256(payload_bytes).hexdigest()

                # 2. Executa a assinatura física via liboqs
                print(f"[DEBUG] Assinando payload interno com {algoritmo_selecionado}...")

                algoritmos_disponiveis = oqs.get_enabled_sig_mechanisms()
                if algoritmo_selecionado not in algoritmos_disponiveis:
                    raise RuntimeError(f"O algoritmo {algoritmo_selecionado} não está habilitado na liboqs.")

                with oqs.Signature(algoritmo_selecionado) as signer:
                    public_key = signer.generate_keypair()
                    # A assinatura recai estritamente sobre os bytes do artefato interno
                    signature = signer.sign(payload_bytes)

                print(f"[DEBUG] Chave gerada: {len(public_key)} bytes | Assinatura física: {len(signature)} bytes")

                # 3. Metadados para leitura rápida
                pqc_metadata = {
                    "algorithm": algoritmo_selecionado,
                    "standard": "FIPS 204",
                    "implementation": "liboqs-python",
                    "timestamp": utc_timestamp,
                    "payload_size_bytes": len(payload_bytes),
                    "hashes": {
                        "sha256": sha256_hash,
                        "sha3_256": sha3_256_hash
                    }
                }

                # 4. Criação do Envelope Final Exposto
                with zipfile.ZipFile(zip_filepath, 'w', zipfile.ZIP_DEFLATED) as zipf:
                    # Grava o ZIP selado (artefato protegido)
                    zipf.writestr("artefato.zip", payload_bytes)
                    
                    # Grava os binários da assinatura física explicitamente como exigido pelo validador
                    zipf.writestr("signature.sig", signature)
                    zipf.writestr("public_key.pub", public_key)
                    
                    # Grava metadados descritivos
                    zipf.writestr("pqc_metadata.json", json.dumps(pqc_metadata, indent=2))

                print(f"\n[✔️] Envelope Criptográfico criado com sucesso.")
                print(f"📦 Estrutura final validável gravada em: '{zip_filepath}'")
                print("   ├── artefato.zip")
                print("   ├── signature.sig")
                print("   ├── public_key.pub")
                print("   └── pqc_metadata.json")
        
        btn_zip.on_click(gerar_zip)
        display(HTML(html_pqc), widgets.HBox([dropdown_pqc, btn_zip]))
    else:
        print("\n[!] Envelope bloqueado: Falha de governança Zero Trust.")

def continuar_fluxo(acionar_auditor):
    ui_decisao.close()
    with saida_interativa:
        estado_global["auditor_ignorado"] = not acionar_auditor
        if acionar_auditor:
            print("\n🕵️ Acionando Auditor Interno (Lendo referência)...")
            estado_global.update(node_auditor(estado_global))
            print("-" * 80)
            print(ler_artefato(estado_global["revisao_auditor_ref"]))
            print("-" * 80)
        else:
            print("\n[⚠️ AVISO] Auditor ignorado pelo operador humano.")
            
        print("\n🧑‍✈️ Acionando Supervisor para veredito final...")
        estado_global.update(node_supervisor(estado_global))
        print("=" * 80)
        print(ler_artefato(estado_global["decisao_final_ref"]))
        print("=" * 80)
        finalizar_esteira()

btn_sim_auditor.on_click(lambda b: continuar_fluxo(True))
btn_nao_auditor.on_click(lambda b: continuar_fluxo(False))

with saida_interativa:
    print("\n🚦 AÇÃO REQUERIDA (HiTL): O fluxo foi pausado. Deseja que o Auditor valide o artefato gerado?")
    display(ui_decisao)
```

```python
# ============================================================================
# @title CÉLULA 8: Validador Universal Pós-Quântico (FIPS 204)
# Objetivo: Identificar algoritmo PQC, calcular hashes e verificar
#           a integridade/autenticidade de artefatos assinados pelo BOMSenso.
# ============================================================================

import os
import glob
import json
import hashlib
import zipfile
import time
import io

import ipywidgets as widgets
from IPython.display import display, HTML, clear_output

# Importação opcional do liboqs
try:
    import oqs
    LIBOQS_DISPONIVEL = True
except ImportError:
    LIBOQS_DISPONIVEL = False

# =====================================================================
# 1. Configuração
# =====================================================================
DIR_ASSINATURAS = os.path.join("assets", "pqc_signatures")
ALGORITMOS_ML_DSA = {
    "ML-DSA-44": {
        "public_key_bytes": 1312,
        "signature_bytes": 2420,
    },
    "ML-DSA-65": {
        "public_key_bytes": 1952,
        "signature_bytes": 3309,
    },
    "ML-DSA-87": {
        "public_key_bytes": 2592,
        "signature_bytes": 4627,
    }
}

# =====================================================================
# 2. Utilitários
# =====================================================================
def calcular_hashes(conteudo):
    """Calcula múltiplos hashes sobre o payload."""
    return {
        "SHA-256": hashlib.sha256(conteudo).hexdigest(),
        "SHA3-256": hashlib.sha3_256(conteudo).hexdigest()
    }

def identificar_ml_dsa_por_tamanho(chave_publica, assinatura):
    """Fallback para identificação da variante ML-DSA."""
    tamanho_chave = len(chave_publica)
    tamanho_assinatura = len(assinatura)

    for algoritmo, parametros in ALGORITMOS_ML_DSA.items():
        if (
            tamanho_chave == parametros["public_key_bytes"]
            and tamanho_assinatura == parametros["signature_bytes"]
        ):
            return algoritmo
    return None

def identificar_algoritmo(metadata, chave_publica=None, assinatura=None):
    """Identifica o algoritmo preferencialmente pelos metadados."""
    if metadata:
        algoritmo = (
            metadata.get("algorithm")
            or metadata.get("pqc_algorithm")
            or metadata.get("signature_algorithm")
        )
        if algoritmo in ALGORITMOS_ML_DSA:
            return algoritmo

    if chave_publica is not None and assinatura is not None:
        return identificar_ml_dsa_por_tamanho(chave_publica, assinatura)
    return None

def listar_arquivos_zip():
    if not os.path.exists(DIR_ASSINATURAS):
        return ["Nenhum artefato encontrado"]

    arquivos = glob.glob(os.path.join(DIR_ASSINATURAS, "*.zip"))

    if not arquivos:
        return ["Nenhum artefato encontrado"]
    return [os.path.basename(f) for f in sorted(arquivos, reverse=True)]

# =====================================================================
# 3. Interface
# =====================================================================
html_header = """
<div style="padding: 15px; border-left: 5px solid #004d40;
background-color: #1a1a1a; margin-bottom: 15px;">

    <h3 style="margin:0; color:#80cbc4;">
        🛡️ BOMSenso | Verificação de Integridade PQC
    </h3>

    <p style="font-size:13px; color:#ccc; margin-top:5px;">
        Identificação automática do algoritmo PQC, cálculo de hashes
        e validação da assinatura utilizando <b>liboqs</b>.
    </p>

</div>
"""
dropdown_arquivos = widgets.Dropdown(
    options=listar_arquivos_zip(),
    description="📦 Projeto:",
    style={"description_width": "initial"},
    layout=widgets.Layout(width="400px")
)

uploader_externo = widgets.FileUpload(
    accept=".zip",
    multiple=False,
    description="📂 Buscar no PC",
    button_style="warning",
    layout=widgets.Layout(width="150px")
)

btn_validar = widgets.Button(
    description=" Validar Artefato PQC",
    button_style="info",
    icon="check-circle",
    layout=widgets.Layout(width="250px")
)

btn_limpar = widgets.Button(
    description=" Limpar Upload",
    icon="trash",
    layout=widgets.Layout(width="150px")
)

output_validacao = widgets.Output()

# =====================================================================
# 4. Extração do Envelope PQC (Leitura Real dos Binários)
# =====================================================================
def extrair_envelope(conteudo_zip):
    with zipfile.ZipFile(io.BytesIO(conteudo_zip)) as z:
        arquivos = z.namelist()
        
        # 1. Lê o artefato interno protegido (Payload Real)
        if "artefato.zip" not in arquivos:
            raise ValueError("O envelope não contém 'artefato.zip'. Este não é um envelope PQC válido do BOMSenso.")
        payload_interno = z.read("artefato.zip")
        
        # 2. Lê a assinatura PQC (suporta .bin ou .sig)
        if "signature.bin" in arquivos:
            signature = z.read("signature.bin")
        elif "signature.sig" in arquivos:
            signature = z.read("signature.sig")
        else:
            raise ValueError("Arquivo de assinatura (signature.bin ou .sig) ausente no envelope.")
            
        # 3. Lê a chave pública PQC (suporta .bin ou .pub)
        if "public_key.bin" in arquivos:
            public_key = z.read("public_key.bin")
        elif "public_key.pub" in arquivos:
            public_key = z.read("public_key.pub")
        else:
            raise ValueError("Arquivo de chave pública (public_key.bin ou .pub) ausente no envelope.")
            
        # 4. Lê os Metadados
        if "pqc_metadata.json" in arquivos:
            metadata = json.loads(z.read("pqc_metadata.json"))
        else:
            metadata = {}

        return {
            "metadata": metadata,
            "payload": payload_interno,
            "signature": signature,
            "public_key": public_key,
            "arquivos": arquivos
        }

# =====================================================================
# 5. Verificação Criptográfica
# =====================================================================
def verificar_assinatura(algoritmo, payload, assinatura, chave_publica):
    if not LIBOQS_DISPONIVEL:
        return None, "liboqs não está disponível neste ambiente."
    try:
        with oqs.Signature(algoritmo, "") as verifier:
            resultado = verifier.verify(
                payload,
                assinatura,
                chave_publica
            )
        return resultado, None
    except Exception as erro:
        return False, str(erro)

# =====================================================================
# 6. Validação Principal
# =====================================================================
def executar_validacao(b):
    with output_validacao:
        clear_output()
        print("🔄 Iniciando validação criptográfica...")
        time.sleep(0.5)

        if uploader_externo.value:
            arquivo_up = (
                uploader_externo.value[0]
                if isinstance(uploader_externo.value, tuple)
                else list(uploader_externo.value.values())[0]
            )
            nome_arquivo = arquivo_up["name"]
            conteudo_zip = arquivo_up["content"]
            print(f"🌐 Artefato EXTERNO: {nome_arquivo}")
            fonte = "EXTERNO"
        else:
            nome_arquivo = dropdown_arquivos.value
            if nome_arquivo == "Nenhum artefato encontrado":
                print("❌ Nenhum artefato ZIP disponível.")
                return
            caminho = os.path.join(DIR_ASSINATURAS, nome_arquivo)
            with open(caminho, "rb") as arquivo:
                conteudo_zip = arquivo.read()
            print(f"📂 Artefato INTERNO: {nome_arquivo}")
            fonte = "INTERNO"

        try:
            envelope = extrair_envelope(conteudo_zip)
        except Exception as erro:
            print(f"❌ Falha ao interpretar envelope: {erro}")
            return
            
        payload = envelope["payload"]
        assinatura = envelope["signature"]
        chave_publica = envelope["public_key"]
        metadata = envelope["metadata"]

        algoritmo = identificar_algoritmo(metadata, chave_publica, assinatura)
        if algoritmo is None:
            print("❌ Não foi possível identificar a variante ML-DSA.")
            print(f"   Chave pública: {len(chave_publica)} bytes")
            print(f"   Assinatura: {len(assinatura)} bytes")
            return
        print(f"🔐 Algoritmo identificado: {algoritmo}")

        print("🧮 Calculando hashes do artefato interno...")
        hashes = calcular_hashes(payload)
        print()
        print("HASH DO ARTEFATO PROTEGIDO")
        print("-" * 60)
        for algoritmo_hash, digest in hashes.items():
            print(f"{algoritmo_hash:<12}: {digest}")

        print()
        print("INFORMAÇÕES CRIPTOGRÁFICAS")
        print("-" * 60)
        print(f"Fonte              : {fonte}")
        print(f"Payload            : {len(payload):,} bytes")
        print(f"Chave pública      : {len(chave_publica):,} bytes")
        print(f"Assinatura         : {len(assinatura):,} bytes")

        if metadata:
            print(f"Versão PQC         : {metadata.get('standard', 'N/A')}")
            print(f"Implementação      : {metadata.get('implementation', 'N/A')}")

        print()
        print("🔎 Verificando assinatura com liboqs...")
        resultado, erro = verificar_assinatura(algoritmo, payload, assinatura, chave_publica)
        
        if erro:
            print(f"⚠️ Verificação criptográfica não executada: {erro}")
            return
            
        if resultado:
            html_sucesso = f"""
            <div style="margin-top:15px; padding:15px; background-color:#1b5e20; color:white; border-radius:4px;">
                <strong>[STATUS: ÍNTEGRO]</strong><br><br>
                Artefato {fonte} validado com sucesso.<br>
                <b>Algoritmo:</b> {algoritmo}<br>
                <b>SHA-256:</b> {hashes["SHA-256"]}<br><br>
                A assinatura ML-DSA foi validada criptograficamente pela liboqs contra o artefato original interno.
            </div>
            """
            display(HTML(html_sucesso))
        else:
            display(HTML("""
            <div style="margin-top:15px; padding:15px; background-color:#b71c1c; color:white; border-radius:4px;">
                <strong>[STATUS: INVÁLIDO]</strong><br><br>
                A assinatura PQC não corresponde ao payload apresentado. A cadeia de integridade deve ser considerada comprometida.
            </div>
            """))

# =====================================================================
# 7. Limpeza
# =====================================================================
def limpar_upload(b):
    uploader_externo.value = ()
    with output_validacao:
        clear_output()
        print("🧹 Upload removido.")

btn_validar.on_click(executar_validacao)
btn_limpar.on_click(limpar_upload)

# =====================================================================
# 8. Renderização
# =====================================================================
display(HTML(html_header))
display(widgets.HBox([dropdown_arquivos, uploader_externo, btn_limpar]))
display(widgets.HBox([btn_validar]))
display(output_validacao)
```
